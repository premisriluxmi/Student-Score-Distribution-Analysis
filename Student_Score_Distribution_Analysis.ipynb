```python
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

```python
scores = [
    42, 45, 48, 50, 52,
    54, 55, 56, 58, 59,
    60, 61, 62, 63, 64,
    65, 66, 67, 68, 69,
    70, 72, 73, 74, 75,
    77, 79, 82, 86, 91
]

df = pd.DataFrame({
    "Student_ID": range(1, 31),
    "Score": scores
})

df.to_csv("student_scores.csv", index=False)

print("Dataset created successfully!")
print("Total students:", len(df))

df.head()
```

```python
print("Dataset Information:")
print(df.info())

print("\nFirst 10 Student Records:")
print(df.head(10))

print("\nStatistical Summary:")
print(df["Score"].describe())
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["Score"],
    bins=6,
    edgecolor="black"
)

plt.title("Distribution of Student Scores")
plt.xlabel("Student Score")
plt.ylabel("Number of Students")
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.savefig("student_score_histogram.png", dpi=300)
plt.show()
```

```python
mean_score = df["Score"].mean()
median_score = df["Score"].median()
min_score = df["Score"].min()
max_score = df["Score"].max()
score_range = max_score - min_score

print("DISTRIBUTION ANALYSIS")
print("---------------------")
print(f"Mean Score: {mean_score:.2f}")
print(f"Median Score: {median_score:.2f}")
print(f"Minimum Score: {min_score}")
print(f"Maximum Score: {max_score}")
print(f"Score Range: {score_range}")

print("\nOBSERVATIONS")
print("1. Scores are spread across different score intervals.")
print("2. The minimum score is", min_score,
      "and the maximum score is", max_score, ".")
print("3. The histogram helps identify score concentration.")
print("4. Comparing the mean and median helps us examine possible skewness.")
```

```python
skewness = df["Score"].skew()

print(f"Skewness: {skewness:.2f}")

if skewness > 0.5:
    print("The distribution is positively skewed (right-skewed).")
elif skewness < -0.5:
    print("The distribution is negatively skewed (left-skewed).")
else:
    print("The distribution is approximately symmetric or has only mild skewness.")
```
